```python
# ============================================================
# ONLINE RETAIL SALES & CUSTOMER BEHAVIOR ANALYSIS
# Complete Data Science + Machine Learning Project
# ============================================================


# 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
import warnings

warnings.filterwarnings("ignore")

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from scipy.cluster.hierarchy import dendrogram, linkage

print("Libraries imported successfully!")


# 2. Upload Dataset

from google.colab import files

uploaded = files.upload()

print("Dataset uploaded successfully!")


# 3. Load Dataset

file_name = list(uploaded.keys())[0]

if file_name.endswith(".xlsx"):
    df = pd.read_excel(file_name)

elif file_name.endswith(".csv"):
    df = pd.read_csv(file_name)

else:
    raise ValueError("Please upload an Excel or CSV file.")

print("Dataset Loaded Successfully!")
print("File Name:", file_name)
print("Shape:", df.shape)


# 4. View Dataset

print("First 5 Rows:")
display(df.head())


# 5. Dataset Information

print("Dataset Information:")
df.info()


# 6. Missing Values

print("Missing Values:")
print(df.isnull().sum())


# 7. Duplicate Records

print("Duplicate Rows:", df.duplicated().sum())


# 8. Statistical Summary

display(df.describe())


# 9. Convert Invoice Date

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"]
)

print("InvoiceDate converted successfully!")


# 10. Create Total Amount

df["TotalAmount"] = (
    df["Quantity"] * df["UnitPrice"]
)

print("TotalAmount column created!")


# 11. Data Cleaning

df = df.drop_duplicates()

df = df.dropna(
    subset=["Description"]
)

df = df.dropna(
    subset=["CustomerID"]
)

df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
]

print("Data Cleaning Completed!")
print("Cleaned Dataset Shape:", df.shape)


# 12. Check Cleaned Data

print("Remaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicates:")
print(df.duplicated().sum())


# 13. Top Products by Revenue

top_products = (
    df.groupby("Description")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_products)


# 14. Top Countries by Revenue

country_sales = (
    df.groupby("Country")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(country_sales)


# 15. Unique Customers

total_customers = df["CustomerID"].nunique()

print("Total Customers:", total_customers)


# 16. Unique Products

total_products = df["StockCode"].nunique()

print("Total Products:", total_products)


# 17. Create Month Column

df["Month"] = (
    df["InvoiceDate"]
    .dt.to_period("M")
)

print("Month column created!")


# 18. Monthly Sales

monthly_sales = (
    df.groupby("Month")["TotalAmount"]
    .sum()
)

display(monthly_sales)


# 19. Monthly Revenue Visualization

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()


# 20. RFM Reference Date

reference_date = (
    df["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

print("Reference Date:", reference_date)


# 21. RFM Analysis

rfm = df.groupby("CustomerID").agg({

    "InvoiceDate":
        lambda x:
        (reference_date - x.max()).days,

    "InvoiceNo":
        "nunique",

    "TotalAmount":
        "sum"
})

rfm.columns = [
    "Recency",
    "Frequency",
    "Monetary"
]

print("RFM Analysis Completed!")

display(rfm.head())


# 22. RFM Summary

print("RFM Summary:")
display(
    rfm[
        ["Recency", "Frequency", "Monetary"]
    ].describe()
)


# 23. Scale RFM Data

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm[
        ["Recency", "Frequency", "Monetary"]
    ]
)

print("RFM Scaling Completed!")


# 24. K-Means Clustering

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = (
    kmeans.fit_predict(rfm_scaled)
)

print("K-Means Clustering Completed!")

print("\nCluster Counts:")
print(
    rfm["Cluster"]
    .value_counts()
    .sort_index()
)


# 25. Hierarchical Clustering Dendrogram

sample_rfm = rfm_scaled[:300]

linked = linkage(
    sample_rfm,
    method="ward"
)

plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title(
    "Hierarchical Clustering Dendrogram"
)

plt.xlabel("Customers")
plt.ylabel("Distance")

plt.show()


# 26. Hierarchical Clustering

hierarchical = AgglomerativeClustering(
    n_clusters=4
)

rfm["Hierarchical_Cluster"] = (
    hierarchical.fit_predict(rfm_scaled)
)

print("Hierarchical Clustering Completed!")

print("\nHierarchical Cluster Counts:")
print(
    rfm["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)


# 27. Cluster Profile

cluster_profile = (
    rfm.groupby("Cluster")
    [["Recency", "Frequency", "Monetary"]]
    .mean()
    .round(2)
)

print("Cluster Profile:")
display(cluster_profile)


# 28. Temporal Data Split

cutoff_date = (
    df["InvoiceDate"].max()
    - pd.Timedelta(days=30)
)

feature_data = df[
    df["InvoiceDate"] <= cutoff_date
].copy()

future_data = df[
    df["InvoiceDate"] > cutoff_date
].copy()

print("Cutoff Date:", cutoff_date)
print("Feature Data Shape:", feature_data.shape)
print("Future Data Shape:", future_data.shape)


# 29. Customer Feature Engineering

customer_features = feature_data.groupby(
    "CustomerID"
).agg({

    "InvoiceNo":
        "nunique",

    "Quantity":
        "sum",

    "UnitPrice":
        "mean",

    "TotalAmount":
        ["sum", "mean"],

    "InvoiceDate":
        ["min", "max"]
})


customer_features.columns = [

    "PurchaseFrequency",
    "TotalQuantity",
    "AvgUnitPrice",
    "TotalSpend",
    "AvgTransactionValue",
    "FirstPurchase",
    "LastPurchase"
]


# 30. Create Recency Feature

customer_features["Recency"] = (
    cutoff_date -
    customer_features["LastPurchase"]
).dt.days


# 31. Create Customer Lifetime Feature

customer_features[
    "CustomerLifetimeDays"
] = (
    customer_features["LastPurchase"]
    - customer_features["FirstPurchase"]
).dt.days

display(
    customer_features.head()
)


# 32. Create Future Purchase Target

future_customers = set(
    future_data["CustomerID"].unique()
)

customer_features[
    "Purchased_Future"
] = (
    customer_features.index
    .isin(future_customers)
).astype(int)

print(
    "Future Purchase Target Created!"
)

print(
    customer_features[
        "Purchased_Future"
    ].value_counts()
)


# 33. Prepare ML Features

X = customer_features.drop(
    columns=[
        "Purchased_Future",
        "FirstPurchase",
        "LastPurchase"
    ]
)

y = customer_features[
    "Purchased_Future"
]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)


# 34. Train Test Split

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)


# 35. Logistic Regression Baseline

log_model = LogisticRegression(
    max_iter=1000
)

log_model.fit(
    X_train,
    y_train
)

log_pred = log_model.predict(
    X_test
)

print(
    "Logistic Regression Completed!"
)


# 36. Logistic Regression Evaluation

log_accuracy = accuracy_score(
    y_test,
    log_pred
)

log_precision = precision_score(
    y_test,
    log_pred
)

log_recall = recall_score(
    y_test,
    log_pred
)

log_f1 = f1_score(
    y_test,
    log_pred
)

print("\nLogistic Regression Results")

print("Accuracy :", log_accuracy)
print("Precision:", log_precision)
print("Recall   :", log_recall)
print("F1 Score :", log_f1)


# 37. Random Forest Model

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(
    X_train,
    y_train
)

rf_pred = rf_model.predict(
    X_test
)

print(
    "Random Forest Completed!"
)


# 38. Random Forest Evaluation

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred
)

rf_recall = recall_score(
    y_test,
    rf_pred
)

rf_f1 = f1_score(
    y_test,
    rf_pred
)

print("\nRandom Forest Results")

print("Accuracy :", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1 Score :", rf_f1)


# 39. Model Comparison

model_comparison = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],

    "Accuracy": [
        log_accuracy,
        rf_accuracy
    ],

    "Precision": [
        log_precision,
        rf_precision
    ],

    "Recall": [
        log_recall,
        rf_recall
    ],

    "F1 Score": [
        log_f1,
        rf_f1
    ]
})

display(model_comparison)


# 40. GridSearchCV Hyperparameter Tuning

param_grid = {

    "n_estimators": [
        100,
        200
    ],

    "max_depth": [
        None,
        5,
        10
    ],

    "min_samples_split": [
        2,
        5
    ]
}


grid_rf = GridSearchCV(

    RandomForestClassifier(
        random_state=42
    ),

    param_grid,

    cv=5,

    scoring="f1",

    n_jobs=-1
)


grid_rf.fit(
    X_train,
    y_train
)

print("GridSearchCV Completed!")

print("\nBest Parameters:")
print(
    grid_rf.best_params_
)

print(
    "\nBest Cross-Validation F1:",
    grid_rf.best_score_
)


# 41. Tuned Random Forest

tuned_rf = (
    grid_rf.best_estimator_
)

tuned_rf_pred = (
    tuned_rf.predict(X_test)
)


# 42. Tuned Random Forest Evaluation

tuned_accuracy = accuracy_score(
    y_test,
    tuned_rf_pred
)

tuned_precision = precision_score(
    y_test,
    tuned_rf_pred
)

tuned_recall = recall_score(
    y_test,
    tuned_rf_pred
)

tuned_f1 = f1_score(
    y_test,
    tuned_rf_pred
)

print("\nTuned Random Forest Results")

print("Accuracy :", tuned_accuracy)
print("Precision:", tuned_precision)
print("Recall   :", tuned_recall)
print("F1 Score :", tuned_f1)


# 43. Overfitting Check

train_pred = tuned_rf.predict(
    X_train
)

train_f1 = f1_score(
    y_train,
    train_pred
)

test_f1 = f1_score(
    y_test,
    tuned_rf_pred
)

print("Train F1 Score:", train_f1)
print("Test F1 Score :", test_f1)
print("F1 Gap        :", train_f1 - test_f1)


# 44. Controlled Hyperparameter Tuning

param_grid_controlled = {

    "n_estimators": [
        100,
        200
    ],

    "max_depth": [
        3,
        5,
        7,
        10
    ],

    "min_samples_split": [
        5,
        10
    ],

    "min_samples_leaf": [
        2,
        5
    ]
}


grid_rf_controlled = GridSearchCV(

    RandomForestClassifier(
        random_state=42
    ),

    param_grid_controlled,

    cv=5,

    scoring="f1",

    n_jobs=-1
)


grid_rf_controlled.fit(
    X_train,
    y_train
)

print(
    "Controlled Tuning Completed!"
)

print("\nControlled Best Parameters:")
print(
    grid_rf_controlled.best_params_
)

print(
    "\nControlled Best CV F1:",
    grid_rf_controlled.best_score_
)


# 45. Controlled Model Evaluation

controlled_rf = (
    grid_rf_controlled.best_estimator_
)

controlled_pred = (
    controlled_rf.predict(X_test)
)

controlled_f1 = f1_score(
    y_test,
    controlled_pred
)

print(
    "Controlled Test F1:",
    controlled_f1
)


# 46. Select Final Model

final_model = tuned_rf

final_pred = final_model.predict(
    X_test
)

print(
    "Final Model: Tuned Random Forest"
)


# 47. Final Model Performance

final_accuracy = accuracy_score(
    y_test,
    final_pred
)

final_precision = precision_score(
    y_test,
    final_pred
)

final_recall = recall_score(
    y_test,
    final_pred
)

final_f1 = f1_score(
    y_test,
    final_pred
)

print("\nFINAL MODEL PERFORMANCE")

print("Accuracy :", final_accuracy)
print("Precision:", final_precision)
print("Recall   :", final_recall)
print("F1 Score :", final_f1)


# 48. Confusion Matrix

cm = confusion_matrix(
    y_test,
    final_pred
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot()

plt.title(
    "Tuned Random Forest - Confusion Matrix"
)

plt.show()


# 49. Feature Importance

feature_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance":
        final_model.feature_importances_

})

feature_importance = (
    feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
)

display(
    feature_importance
)


# 50. Feature Importance Visualization

top_features = (
    feature_importance.head(10)
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.title(
    "Top 10 Random Forest Feature Importances"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()


# 51. Create ML Dashboard Data

ml_dashboard = (
    customer_features.copy()
)

ml_dashboard[
    "Predicted_Future_Purchase"
] = final_model.predict(X)

ml_dashboard[
    "Purchase_Probability"
] = final_model.predict_proba(X)[:, 1]

display(
    ml_dashboard.head()
)


# 52. Add Customer Clusters

ml_dashboard = ml_dashboard.join(
    rfm[
        [
            "Cluster",
            "Hierarchical_Cluster"
        ]
    ],
    how="left"
)

display(
    ml_dashboard.head()
)


# 53. Prediction Labels

ml_dashboard[
    "Prediction_Label"
] = (
    ml_dashboard[
        "Predicted_Future_Purchase"
    ]
    .map({
        1: "Predicted Purchase",
        0: "Not Predicted"
    })
)

display(
    ml_dashboard.head()
)


# 54. Prediction Summary

prediction_summary = pd.DataFrame({

    "Prediction": [
        "Predicted Purchase",
        "Not Predicted"
    ],

    "Customers": [

        (
            ml_dashboard[
                "Predicted_Future_Purchase"
            ] == 1
        ).sum(),

        (
            ml_dashboard[
                "Predicted_Future_Purchase"
            ] == 0
        ).sum()

    ]
})

display(
    prediction_summary
)


# 55. Model Performance Data

model_performance = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],

    "Logistic Regression": [
        log_accuracy,
        log_precision,
        log_recall,
        log_f1
    ],

    "Random Forest": [
        rf_accuracy,
        rf_precision,
        rf_recall,
        rf_f1
    ],

    "Tuned Random Forest": [
        tuned_accuracy,
        tuned_precision,
        tuned_recall,
        tuned_f1
    ]
})

display(
    model_performance
)


# 56. Create SQLite Database

conn = sqlite3.connect(
    "online_retail.db"
)

df_sql = df.copy()

df_sql["Month"] = (
    df_sql["Month"].astype(str)
)

df_sql.to_sql(
    "retail",
    conn,
    if_exists="replace",
    index=False
)

print(
    "SQLite Database Created!"
)


# 57. SQL - Top Customers

query = """

SELECT
    CustomerID,
    SUM(TotalAmount) AS Total_Spend

FROM retail

GROUP BY CustomerID

ORDER BY Total_Spend DESC

LIMIT 10

"""

top_customers_sql = (
    pd.read_sql_query(
        query,
        conn
    )
)

display(
    top_customers_sql
)


# 58. SQL - Top Products

query = """

SELECT
    Description,
    SUM(TotalAmount) AS Total_Revenue

FROM retail

GROUP BY Description

ORDER BY Total_Revenue DESC

LIMIT 10

"""

top_products_sql = (
    pd.read_sql_query(
        query,
        conn
    )
)

display(
    top_products_sql
)


# 59. SQL - Monthly Revenue

query = """

SELECT
    Month,
    SUM(TotalAmount) AS Monthly_Revenue

FROM retail

GROUP BY Month

ORDER BY Month

"""

monthly_revenue_sql = (
    pd.read_sql_query(
        query,
        conn
    )
)

display(
    monthly_revenue_sql
)


# 60. SQL - Country Revenue

query = """

SELECT
    Country,
    SUM(TotalAmount) AS Total_Revenue

FROM retail

GROUP BY Country

ORDER BY Total_Revenue DESC

LIMIT 10

"""

country_revenue_sql = (
    pd.read_sql_query(
        query,
        conn
    )
)

display(
    country_revenue_sql
)


# 61. SQL - Purchase Frequency

query = """

SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo)
        AS Purchase_Frequency

FROM retail

GROUP BY CustomerID

ORDER BY Purchase_Frequency DESC

LIMIT 10

"""

purchase_frequency_sql = (
    pd.read_sql_query(
        query,
        conn
    )
)

display(
    purchase_frequency_sql
)


# 62. Business Metrics

total_revenue = (
    df["TotalAmount"].sum()
)

total_customers = (
    df["CustomerID"].nunique()
)

total_products = (
    df["StockCode"].nunique()
)

total_orders = (
    df["InvoiceNo"].nunique()
)

average_order_value = (
    total_revenue /
    total_orders
)

top_country = (
    df.groupby("Country")["TotalAmount"]
    .sum()
    .idxmax()
)

top_product = (
    df.groupby("Description")["TotalAmount"]
    .sum()
    .idxmax()
)

print("Total Revenue:", total_revenue)
print("Total Customers:", total_customers)
print("Total Products:", total_products)
print("Total Orders:", total_orders)
print("Average Order Value:", average_order_value)
print("Top Country:", top_country)
print("Top Product:", top_product)


# 63. Export Main Power BI Data

df.to_csv(
    "retail_powerbi.csv",
    index=False
)

print(
    "retail_powerbi.csv created!"
)


# 64. Export ML Dashboard Data

ml_dashboard_export = (
    ml_dashboard.reset_index()
)

ml_dashboard_export.to_csv(
    "ml_dashboard.csv",
    index=False
)

print(
    "ml_dashboard.csv created!"
)


# 65. Export Model Performance

model_performance.to_csv(
    "model_performance.csv",
    index=False
)

print(
    "model_performance.csv created!"
)


# 66. Export Prediction Summary

prediction_summary.to_csv(
    "prediction_summary.csv",
    index=False
)

print(
    "prediction_summary.csv created!"
)


# 67. Export Cluster Profile

cluster_profile.to_csv(
    "cluster_profile.csv"
)

print(
    "cluster_profile.csv created!"
)


# 68. Close Database

conn.close()

print(
    "Database connection closed!"
)


# 69. Final Project Summary

print("\nONLINE RETAIL SALES & CUSTOMER BEHAVIOR ANALYSIS")
print("------------------------------------------------")

print("Data Loading        : Completed")
print("Data Cleaning       : Completed")
print("EDA                 : Completed")
print("RFM Analysis        : Completed")
print("K-Means             : Completed")
print("Hierarchical        : Completed")
print("Logistic Regression : Completed")
print("Random Forest       : Completed")
print("GridSearchCV        : Completed")
print("Overfitting Check   : Completed")
print("Feature Importance  : Completed")
print("SQL Analysis        : Completed")
print("Business Metrics    : Completed")
print("Power BI Export     : Completed")
print("ML Dashboard Data   : Completed")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
